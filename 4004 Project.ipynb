```python
import pandas as pd
df = pd.read_excel('DOHMH_HIV_AIDS_Annual_Report_20261027.xlsx')
df
```

```python
df.info()
```

```python
# use 2020,2021,2022 to average indicators

df = df[df['Year'].isin([2020, 2021, 2022])]
df.info()

df
```

```python
#df = df[df['UHF'] != 'All']

#df

#print(df['UHF'].unique())
```

```python
#data contains rows represent the entire city, each boroughs, and each UHF .

df_total = df[
    (df['UHF'] != 'All') &
    (df['Gender'] == 'All') &
    (df['Age'] == 'All') &
    (df['Race'] == 'All')
]

df_total

df_total.info()
```

```python
print(df_total.shape)
print(df_total['UHF'].nunique())
print(df_total.head())
```

```python
print(df_total.isnull().sum())
# not much use pandas replace by mean
```

```python
#average HIV diagnosis rate:
HIV_avg = df_total.groupby(['Borough', 'UHF'])['HIV diagnosis rate'].mean()

print(HIV_avg)
print(HIV_avg.shape)
```

```python
#average care-linkage rate
care_avg = df_total.groupby(
    ['Borough', 'UHF']
)['% linked to care within 3 months'].mean()

print(care_avg)
print(care_avg.shape)
```

```python
df_avg = pd.merge(HIV_avg, care_avg, on=['Borough', 'UHF'])

df_avg
```

```python
#Care Linkage Gap:

df_avg['Care Linkage Gap'] = 1 - df_avg['% linked to care within 3 months']

df_avg
```

```python
#racial disparity

df_race = df[
    (df['Race'].isin(['Black', 'Latinx/Hispanic'])) &
    (df['Gender'] == 'All') &
    (df['Age'] == 'All') &
    (df['UHF'] != 'All')
]

df_race
```

```python
print(df_race[['Year', 'UHF', 'Race', 'HIV diagnoses', 'HIV diagnosis rate']])
```

```python
# population count

import numpy as np

df_race['Population'] = (
    df_race['HIV diagnoses'] /
    df_race['HIV diagnosis rate'] * 100000
)

df_race
```

```python
# black and hispanic population
race_total = df_race.groupby(['Borough', 'UHF', 'Year'])[['HIV diagnoses', 'Population']].sum().reset_index()
race_total
```

```python
race_total['Racial Combine Rate'] = (race_total['HIV diagnoses'] /race_total['Population'] * 100000)
race_total

#new HIV diagnoses rate per 100,000 Black and Latino residents
```

```python
race_avg_rate = race_total.groupby(['Borough', 'UHF'])['Racial Combine Rate'].mean()
race_avg_rate
```

```python
print(race_avg_rate.shape)
```

```python
#nyc racial diagnose rate

# cal nyc population using HIV rate

df_total['Population'] = (
    df_total['HIV diagnoses'] /
    df_total['HIV diagnosis rate'] * 100000
)

df_total
```

```python
nyc_total = df_total.groupby('Year')[
    ['HIV diagnoses', 'Population']
].sum()

nyc_total['NYC Rate'] = (
    nyc_total['HIV diagnoses'] /
    nyc_total['Population'] * 100000
)

nyc_total
```

```python
race_total = race_total.merge(
    nyc_total[['NYC Rate']],
    on='Year'
)
race_total['Racial Disparity'] = (
    race_total['Racial Combine Rate'] / race_total['NYC Rate']
)
race_total
```

```python
race_avg = race_total.groupby(['Borough', 'UHF'])['Racial Disparity'].mean()

race_avg
```

```python
df_avg = df_avg.merge(race_avg, on=['Borough', 'UHF'])

df_avg
```

```python
# normalize varibale
df_avg['R_norm'] = (
    df_avg['HIV diagnosis rate'] /
    df_avg['HIV diagnosis rate'].max()
)

df_avg['RD_norm'] = (
    df_avg['Racial Disparity'] /
    df_avg['Racial Disparity'].max()
)

df_avg['CG_norm'] = (
    df_avg['Care Linkage Gap'] /
    df_avg['Care Linkage Gap'].max()
)

df_avg
```

```python
df_avg['Need Score'] = (
    0.45 * df_avg['R_norm'] +
    0.35 * df_avg['RD_norm'] +
    0.20 * df_avg['CG_norm']
)

df_avg
```

```python
# PLWDHI prevalence approximately # out of every 100 residents are living with diagnosed HIV.

df_total['Pn'] = (df_total['PLWDHI prevalence'] / 100* df_total['Population'])

df_total.info()
```

```python
df_total['Unmet Need'] = (df_total['Pn'] * (1 - df_total['% viral suppression']))
df_total[['Year', 'UHF', 'Pn', 'Unmet Need']]
```

```python
unmet_avg = df_total.groupby(['Borough', 'UHF'])['Unmet Need'].mean().reset_index()
unmet_avg['Funding Need'] = unmet_avg['Unmet Need'] * 3400
unmet_avg
```

```python
df_avg = df_avg.merge(unmet_avg, on=['Borough', 'UHF'])

df_avg

```
